```python
import torch
```

# casting a tensor

```python
# using to function
t1=torch.linspace(1,10,10)
t1
print(f"orginal_tensor:{t1},data type:{t1.dtype}")
casted_tensor=t1.to(torch.int32)
print(f"casted_tensor:{casted_tensor}")
```

```python
# using type function
cpu_tensor=torch.randn(3,3)
f16_tensor=cpu_tensor.type(torch.float16)
print(f"cpu_tensor:{cpu_tensor},data type:{cpu_tensor.dtype}")
print(f"f16_tensor:{f16_tensor},data type:{f16_tensor.dtype}")
```
