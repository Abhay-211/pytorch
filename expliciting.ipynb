```python
import torch
```

```python
# Default is usually float32 for floating point numbers
tensor_float=torch.tensor([1.0,2.0,3.0])
print(f"default float dtype: {tensor_float.dtype}")
```

```python
#Explicitly creating a float64 tensor
t1=torch.tensor([1,2,3],dtype=torch.float64)

print(f"float64 dtype: {t1.dtype}")
```

```python
# creating an integer tensor
t2=torch.tensor([1,2,3],dtype=torch.int32)
t2
```

```python
boolean=torch.tensor([True,False,True],dtype=torch.bool)
print(f"Boolean tensor dtype: {boolean.dtype}")
print(boolean)
```
