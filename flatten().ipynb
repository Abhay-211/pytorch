```python
import torch
```

```python
# flatten the entire tensor
t=torch.arange(0,10).reshape(2,5)

print(f' normal tensor:{t}')
ft=torch.flatten(t)
print(f'flattened tensor:{ft}')
```

```python
t3=torch.randn(2,3,4)
t3
flattened_part=torch.flatten(t3,start_dim=1,end_dim=1)
print(flattened_part)
```
